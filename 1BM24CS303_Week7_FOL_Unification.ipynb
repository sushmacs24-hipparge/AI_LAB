```python
def occurs(var, term):
    if var == term:
        return True
    if isinstance(term, tuple):
        return any(occurs(var, arg) for arg in term[1])
    return False


def apply_subst(subst, term):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, tuple):
        return (term[0], tuple(apply_subst(subst, arg) for arg in term[1]))

    return term


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = apply_subst(subst, x)
    y = apply_subst(subst, y)

    if x == y:
        return subst

    if isinstance(x, str):
        if occurs(x, y):
            return None
        subst[x] = y
        return subst

    if isinstance(y, str):
        if occurs(y, x):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x[1]) != len(y[1]):
            return None

        for a, b in zip(x[1], y[1]):
            subst = unify(a, b, subst)
            if subst is None:
                return None

        return subst

    return None


def parse_expression(s):
    name, args = s.split("(", 1)
    args = args.rstrip(")").split(",")
    return (name.strip(), tuple(arg.strip() for arg in args))


e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

x = parse_expression(e1)
y = parse_expression(e2)

result = unify(x, y)

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
```
