```python
def pl_true(sentence, model):
    if isinstance(sentence, str):
        return model[sentence]

    op = sentence[0]

    if op == 'NOT':
        return not pl_true(sentence[1], model)
    elif op == 'AND':
        return pl_true(sentence[1], model) and pl_true(sentence[2], model)
    elif op == 'OR':
        return pl_true(sentence[1], model) or pl_true(sentence[2], model)
    elif op == 'IMPLIES':
        return (not pl_true(sentence[1], model)) or pl_true(sentence[2], model)

def generate_models(symbols, model={}):
    if not symbols:
        return [model]
    P = symbols[0]
    rest = symbols[1:]
    m_false = model.copy()
    m_false[P] = False
    m_true = model.copy()
    m_true[P] = True
    return generate_models(rest, m_false) + generate_models(rest, m_true)

kb = ('AND',
        ('IMPLIES', 'Q', 'P'),
        ('AND',
            ('IMPLIES', 'P', ('NOT', 'Q')),
            ('OR', 'Q', 'R')
        )
     )

q1 = 'R'
q2 = ('IMPLIES', 'R', 'P')
q3 = ('IMPLIES', 'Q', 'R')

symbols = ['P', 'Q', 'R']
models = generate_models(symbols)

s1 = ('IMPLIES', 'Q', 'P')
s2 = ('IMPLIES', 'P', ('NOT', 'Q'))
s3 = ('OR', 'Q', 'R')

headers = ["P", "Q", "R", "Q->P", "P->~Q", "Q|R", "KB", "R", "R->P", "Q->R"]
print(" | ".join(f"{h:^7}" for h in headers))
print("-" * 85)

kb_models = []

for m in models:
    v_s1 = pl_true(s1, m)
    v_s2 = pl_true(s2, m)
    v_s3 = pl_true(s3, m)
    v_kb = pl_true(kb, m)

    v_q1 = pl_true(q1, m)
    v_q2 = pl_true(q2, m)
    v_q3 = pl_true(q3, m)

    if v_kb:
        kb_models.append(m)

    row = [
        str(m['P']), str(m['Q']), str(m['R']),
        str(v_s1), str(v_s2), str(v_s3),
        str(v_kb), str(v_q1), str(v_q2), str(v_q3)
    ]
    print(" | ".join(f"{val:^7}" for val in row))

print("-" * 85)

e1 = all(pl_true(q1, m) for m in kb_models)
e2 = all(pl_true(q2, m) for m in kb_models)
e3 = all(pl_true(q3, m) for m in kb_models)

print(f"Does KB entail R? {e1}")
print(f"Does KB entail R -> P? {e2}")
print(f"Does KB entail Q -> R? {e3}")
```
